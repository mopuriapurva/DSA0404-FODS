```python
# ==========================================
# CO3 - AT1 Case Study Solutions
# Google Colab Code
# ==========================================

import numpy as np
import statistics as st

# ------------------------------------------
# Helper Functions
# ------------------------------------------

def descriptive_stats(data):
    data = np.array(data)

    print("Data:", data)

    print("Mean =", np.mean(data))
    print("Median =", np.median(data))

    try:
        print("Mode =", st.mode(data))
    except:
        print("Mode = No unique mode")

    print("Range =", np.max(data) - np.min(data))

    print("Variance =", np.var(data))
    print("Standard Deviation =", np.std(data))

def quartiles(data):
    data = np.array(sorted(data))
    q1 = np.percentile(data,25)
    q2 = np.percentile(data,50)
    q3 = np.percentile(data,75)

    iqr = q3-q1

    lower = q1-1.5*iqr
    upper = q3+1.5*iqr

    outliers = data[(data<lower)|(data>upper)]

    print("Q1 =",q1)
    print("Q2 =",q2)
    print("Q3 =",q3)
    print("IQR =",iqr)
    print("Outliers =",outliers)

# ==========================================
# Case Study 1
# ==========================================

print("\n==============================")
print("CASE STUDY 1")
print("==============================")

salary=[4,5,5,6,7,8,8,9,30]

descriptive_stats(salary)
quartiles(salary)

# ==========================================
# Case Study 2
# ==========================================

print("\n==============================")
print("CASE STUDY 2")
print("==============================")

response=[120,125,128,130,132,135,140,145,150,300]

descriptive_stats(response)

print("Skewness : Right Skewed")
print("Median is preferred because of the outlier (300).")

# ==========================================
# Case Study 3
# ==========================================

print("\n==============================")
print("CASE STUDY 3")
print("==============================")

marks=[42,45,48,50,52,52,55,58,60,62,65,95]

quartiles(marks)

print("Mean =",np.mean(marks))
print("Median =",np.median(marks))

# ==========================================
# Case Study 4
# ==========================================

print("\n==============================")
print("CASE STUDY 4")
print("==============================")

orders=[18,20,22,22,24,25,26,28,30,35]

descriptive_stats(orders)

cv=(np.std(orders)/np.mean(orders))*100

print("Coefficient of Variation =",cv,"%")

# ==========================================
# Case Study 5
# ==========================================

print("\n==============================")
print("CASE STUDY 5")
print("==============================")

A=[2,3,3,4,4,5,5,6]
B=[1,1,2,4,5,7,8,9]

print("Model A")
descriptive_stats(A)

print()

print("Model B")
descriptive_stats(B)

if np.std(A)<np.std(B):
    print("Model A is more consistent")
else:
    print("Model B is more consistent")

# ==========================================
# Case Study 6
# ==========================================

print("\n==============================")
print("CASE STUDY 6")
print("==============================")

mean=2000
sd=400
x=3200

z=(x-mean)/sd

print("Z-score =",z)

if z>2:
    print("Customer spending is unusually high")
else:
    print("Customer spending is normal")

print("Negative z-score means spending is below average.")
print("z = -1.5 means customer spends 1.5 SD below average.")

# ==========================================
# Case Study 7
# ==========================================

print("\n==============================")
print("CASE STUDY 7")
print("==============================")

delivery=[25,27,28,29,30,30,31,32,34,60]

descriptive_stats(delivery)
quartiles(delivery)

print("Median better represents delivery time because of outlier.")

# ==========================================
# Case Study 8
# ==========================================

print("\n==============================")
print("CASE STUDY 8")
print("==============================")

cityA=[28,29,30,30,31,30,32]
cityB=[20,25,30,35,30,25,45]

print("City A")
descriptive_stats(cityA)

print()

print("City B")
descriptive_stats(cityB)

if np.var(cityA)<np.var(cityB):
    print("City A has more predictable temperatures.")
else:
    print("City B has more predictable temperatures.")

# ==========================================
# Case Study 9
# ==========================================

print("\n==============================")
print("CASE STUDY 9")
print("==============================")

cpu=[40,42,45,45,46,48,50,52,55,90]

descriptive_stats(cpu)
quartiles(cpu)

print("Skewness : Right Skewed")

# ==========================================
# Case Study 10
# ==========================================

print("\n==============================")
print("CASE STUDY 10")
print("==============================")

A=[10,20,30,40,50]
B=[28,29,30,31,32]

print("Dataset A")
descriptive_stats(A)

print()

print("Dataset B")
descriptive_stats(B)

if np.std(A)<np.std(B):
    print("Dataset A is more consistent")
else:
    print("Dataset B is more consistent")
```
