```python
# ================================================================
#        EXPLORATORY DATA ANALYSIS (EDA) - ALL 8 QUESTIONS
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

print("=" * 70)
print("        EXPLORATORY DATA ANALYSIS - PYTHON PRACTICAL")
print("=" * 70)


# ================================================================
# Q1. COMPLETE SUMMARY OF EMPLOYEE DATASET
# ================================================================

print("\n\n" + "=" * 70)
print("Q1. COMPLETE SUMMARY OF EMPLOYEE DATASET")
print("=" * 70)

employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["John", "Alice", "Bob", "David", "Emma", "Sam"],
    "Age": [25, 30, 28, 35, 29, 40],
    "Department": ["HR", "IT", "IT", "Finance", "HR", "IT"],
    "Salary": [30000, 50000, 45000, 60000, 35000, 70000]
}

employee_df = pd.DataFrame(employee_data)

print("\nEmployee Dataset:")
display(employee_df)

print("\nNumber of Rows:", employee_df.shape[0])
print("Number of Columns:", employee_df.shape[1])

print("\nData Types:")
print(employee_df.dtypes)

print("\nMissing Values:")
print(employee_df.isnull().sum())

print("\nStatistical Summary:")
display(employee_df.describe())


# ================================================================
# Q2. FREQUENCY DISTRIBUTION OF STUDENTS' GRADES
# ================================================================

print("\n\n" + "=" * 70)
print("Q2. FREQUENCY DISTRIBUTION OF STUDENTS' GRADES")
print("=" * 70)

grades = [
    "A", "B", "A", "C", "B", "A",
    "D", "B", "C", "A", "B", "C",
    "A", "D", "B", "A", "C", "B"
]

grade_frequency = pd.Series(grades).value_counts().sort_index()

print("\nFrequency Distribution:")
print(grade_frequency)

plt.figure(figsize=(8, 5))
plt.bar(grade_frequency.index, grade_frequency.values)
plt.xlabel("Grade")
plt.ylabel("Number of Students")
plt.title("Frequency Distribution of Students' Grades")
plt.show()


# ================================================================
# Q3. HISTOGRAM OF HOUSE PRICES
# ================================================================

print("\n\n" + "=" * 70)
print("Q3. HISTOGRAM OF HOUSE PRICES")
print("=" * 70)

house_prices = [
    120000, 150000, 180000, 200000,
    220000, 250000, 275000, 300000,
    320000, 350000, 380000, 400000,
    450000, 500000, 600000, 700000,
    850000
]

house_df = pd.DataFrame({
    "House_Price": house_prices
})

print("\nHouse Price Data:")
display(house_df)

plt.figure(figsize=(8, 5))
plt.hist(
    house_df["House_Price"],
    bins=8,
    edgecolor="black"
)
plt.xlabel("House Price")
plt.ylabel("Frequency")
plt.title("Histogram of House Prices")
plt.show()

skewness = house_df["House_Price"].skew()

print("Skewness:", round(skewness, 2))

if skewness > 0:
    print("Conclusion: Data is positively skewed (Right Skewed).")
elif skewness < 0:
    print("Conclusion: Data is negatively skewed (Left Skewed).")
else:
    print("Conclusion: Data is approximately normally distributed.")


# ================================================================
# Q4. BOX PLOT OF SALARY DATA AND OUTLIERS
# ================================================================

print("\n\n" + "=" * 70)
print("Q4. BOX PLOT OF SALARY DATA AND OUTLIERS")
print("=" * 70)

salary = [
    25000, 28000, 30000, 32000,
    35000, 38000, 40000, 42000,
    45000, 48000, 50000, 52000,
    150000
]

salary_df = pd.DataFrame({
    "Salary": salary
})

print("\nSalary Dataset:")
display(salary_df)

plt.figure(figsize=(8, 5))
plt.boxplot(salary_df["Salary"])
plt.ylabel("Salary")
plt.title("Box Plot of Salary Data")
plt.show()

Q1 = salary_df["Salary"].quantile(0.25)
Q3 = salary_df["Salary"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)

salary_outliers = salary_df[
    (salary_df["Salary"] < lower_bound) |
    (salary_df["Salary"] > upper_bound)
]

print("\nSalary Outliers:")
display(salary_outliers)


# ================================================================
# Q5. DETECT OUTLIERS IN STUDENTS' MARKS USING IQR
# ================================================================

print("\n\n" + "=" * 70)
print("Q5. DETECT OUTLIERS IN STUDENTS' MARKS USING IQR")
print("=" * 70)

marks = [
    45, 50, 55, 60,
    62, 65, 68, 70,
    72, 75, 78, 80,
    82, 85, 95, 150
]

marks_df = pd.DataFrame({
    "Marks": marks
})

print("\nStudent Marks:")
display(marks_df)

Q1 = marks_df["Marks"].quantile(0.25)
Q3 = marks_df["Marks"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)

marks_outliers = marks_df[
    (marks_df["Marks"] < lower_bound) |
    (marks_df["Marks"] > upper_bound)
]

print("\nDetected Outliers:")
display(marks_outliers)


# ================================================================
# Q6. REMOVE OUTLIERS FROM SALES DATASET
# ================================================================

print("\n\n" + "=" * 70)
print("Q6. REMOVE OUTLIERS FROM SALES DATASET")
print("=" * 70)

sales = [
    1000, 1200, 1500, 1800,
    2000, 2200, 2500, 2700,
    3000, 3200, 3500, 3800,
    4000, 4500, 20000
]

sales_df = pd.DataFrame({
    "Sales": sales
})

print("\nOriginal Sales Dataset:")
display(sales_df)

Q1 = sales_df["Sales"].quantile(0.25)
Q3 = sales_df["Sales"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

sales_outliers = sales_df[
    (sales_df["Sales"] < lower_bound) |
    (sales_df["Sales"] > upper_bound)
]

cleaned_sales = sales_df[
    (sales_df["Sales"] >= lower_bound) &
    (sales_df["Sales"] <= upper_bound)
]

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)

print("\nSales Outliers:")
display(sales_outliers)

print("\nCleaned Sales Dataset:")
display(cleaned_sales)

print("Rows Before:", len(sales_df))
print("Rows After:", len(cleaned_sales))


# ================================================================
# Q7. HISTOGRAMS AND BOX PLOTS BEFORE AND AFTER OUTLIER REMOVAL
# ================================================================

print("\n\n" + "=" * 70)
print("Q7. BEFORE AND AFTER OUTLIER REMOVAL")
print("=" * 70)

data = [
    10, 12, 14, 15, 16, 18,
    20, 21, 22, 23, 24, 25,
    26, 28, 30, 100
]

q7_df = pd.DataFrame({
    "Value": data
})

Q1 = q7_df["Value"].quantile(0.25)
Q3 = q7_df["Value"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

q7_cleaned = q7_df[
    (q7_df["Value"] >= lower_bound) &
    (q7_df["Value"] <= upper_bound)
]

print("\nOriginal Dataset:")
display(q7_df)

print("\nCleaned Dataset:")
display(q7_cleaned)

# Histogram BEFORE
plt.figure(figsize=(8, 5))
plt.hist(q7_df["Value"], bins=8, edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram BEFORE Removing Outliers")
plt.show()

# Histogram AFTER
plt.figure(figsize=(8, 5))
plt.hist(q7_cleaned["Value"], bins=8, edgecolor="black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram AFTER Removing Outliers")
plt.show()

# Box plot BEFORE
plt.figure(figsize=(8, 5))
plt.boxplot(q7_df["Value"])
plt.ylabel("Value")
plt.title("Box Plot BEFORE Removing Outliers")
plt.show()

# Box plot AFTER
plt.figure(figsize=(8, 5))
plt.boxplot(q7_cleaned["Value"])
plt.ylabel("Value")
plt.title("Box Plot AFTER Removing Outliers")
plt.show()


# ================================================================
# Q8. COMPLETE EDA USING IRIS DATASET
# ================================================================

print("\n\n" + "=" * 70)
print("Q8. COMPLETE EDA USING IRIS DATASET")
print("=" * 70)

# Load Iris dataset
iris = load_iris()

iris_df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add species
iris_df["species"] = iris.target

print("\nFirst 5 Rows of Iris Dataset:")
display(iris_df.head())

# ------------------------------------------------
# Dataset Information
# ------------------------------------------------

print("\n--- DATASET INFORMATION ---")

print("Number of Rows:", iris_df.shape[0])
print("Number of Columns:", iris_df.shape[1])

print("\nColumn Names:")
print(iris_df.columns.tolist())

print("\nData Types:")
print(iris_df.dtypes)

print("\nComplete Information:")
iris_df.info()

# ------------------------------------------------
# Missing Values
# ------------------------------------------------

print("\n--- MISSING VALUES ---")

print(iris_df.isnull().sum())

print("\nTotal Missing Values:",
      iris_df.isnull().sum().sum())

# ------------------------------------------------
# Descriptive Statistics
# ------------------------------------------------

print("\n--- DESCRIPTIVE STATISTICS ---")

display(iris_df.describe())

# ------------------------------------------------
# Histograms
# ------------------------------------------------

numeric_columns = iris.feature_names

print("\n--- HISTOGRAMS ---")

for column in numeric_columns:

    plt.figure(figsize=(8, 5))

    plt.hist(
        iris_df[column],
        bins=10,
        edgecolor="black"
    )

    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title("Histogram of " + column)

    plt.show()

# ------------------------------------------------
# Box Plots
# ------------------------------------------------

print("\n--- BOX PLOTS ---")

for column in numeric_columns:

    plt.figure(figsize=(8, 5))

    plt.boxplot(iris_df[column])

    plt.ylabel(column)
    plt.title("Box Plot of " + column)

    plt.show()

# ------------------------------------------------
# Outlier Detection
# ------------------------------------------------

print("\n--- OUTLIER DETECTION USING IQR ---")

for column in numeric_columns:

    Q1 = iris_df[column].quantile(0.25)
    Q3 = iris_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = iris_df[
        (iris_df[column] < lower_bound) |
        (iris_df[column] > upper_bound)
    ]

    print("\nColumn:", column)
    print("Q1 =", round(Q1, 2))
    print("Q3 =", round(Q3, 2))
    print("IQR =", round(IQR, 2))
    print("Lower Bound =", round(lower_bound, 2))
    print("Upper Bound =", round(upper_bound, 2))
    print("Number of Outliers =", len(outliers))

    if len(outliers) > 0:
        print("Outlier Values:",
              outliers[column].values)

# ------------------------------------------------
# Remove Outliers
# ------------------------------------------------

print("\n--- REMOVING OUTLIERS ---")

iris_cleaned = iris_df.copy()

for column in numeric_columns:

    Q1 = iris_cleaned[column].quantile(0.25)
    Q3 = iris_cleaned[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    iris_cleaned = iris_cleaned[
        (iris_cleaned[column] >= lower_bound) &
        (iris_cleaned[column] <= upper_bound)
    ]

iris_cleaned = iris_cleaned.reset_index(drop=True)

# ------------------------------------------------
# Display Cleaned Dataset
# ------------------------------------------------

print("\n--- CLEANED IRIS DATASET ---")

display(iris_cleaned)

print("\nOriginal Dataset Shape:",
      iris_df.shape)

print("Cleaned Dataset Shape:",
      iris_cleaned.shape)

print("Rows Removed:",
      len(iris_df) - len(iris_cleaned))

# ------------------------------------------------
# Box Plots After Cleaning
# ------------------------------------------------

print("\n--- BOX PLOTS AFTER REMOVING OUTLIERS ---")

for column in numeric_columns:

    plt.figure(figsize=(8, 5))

    plt.boxplot(iris_cleaned[column])

    plt.ylabel(column)
    plt.title(
        "Box Plot After Removing Outliers - " + column
    )

    plt.show()

# ------------------------------------------------
# Save Cleaned Dataset
# ------------------------------------------------

iris_cleaned.to_csv(
    "iris_cleaned.csv",
    index=False
)

print("\n" + "=" * 70)
print("CLEANED IRIS DATASET SAVED SUCCESSFULLY")
print("File Name: iris_cleaned.csv")
print("=" * 70)


# ================================================================
# DOWNLOAD CLEANED DATASET
# ================================================================

from google.colab import files

files.download("iris_cleaned.csv")

print("\nAll 8 Questions Completed Successfully!")
```
