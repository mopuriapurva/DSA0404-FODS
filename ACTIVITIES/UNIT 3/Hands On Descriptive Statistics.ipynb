```python
# ============================================================
# STATISTICAL CALCULATIONS USING PYTHON
# ============================================================

# 1. Finding Mean Using Python
print("1. FINDING MEAN")
data = [10, 20, 30, 40, 50]

mean_val = sum(data) / len(data)

print("Data:", data)
print("Mean:", mean_val)


# ============================================================
# 2. Finding Median Using Python
print("\n2. FINDING MEDIAN")

data = [40, 10, 30, 50, 20]

data.sort()
n = len(data)

if n % 2 == 0:
    median_val = (data[n // 2 - 1] + data[n // 2]) / 2
else:
    median_val = data[n // 2]

print("Sorted Data:", data)
print("Median:", median_val)


# ============================================================
# 3. Finding Mode Using Python
print("\n3. FINDING MODE")

from collections import Counter

data = [1, 2, 2, 3, 4, 2, 5]

count = Counter(data)
mode_val = count.most_common(1)[0][0]

print("Data:", data)
print("Mode:", mode_val)


# ============================================================
# 4. Finding Range Using Python
print("\n4. FINDING RANGE")

data = [15, 8, 22, 42, 10]

range_val = max(data) - min(data)

print("Data:", data)
print("Maximum:", max(data))
print("Minimum:", min(data))
print("Range:", range_val)


# ============================================================
# 5. Finding Variance Using Python - Population
print("\n5. FINDING POPULATION VARIANCE")

data = [2, 4, 4, 4, 5, 5, 7, 9]

mean_val = sum(data) / len(data)

variance_val = sum(
    (x - mean_val) ** 2 for x in data
) / len(data)

print("Data:", data)
print("Mean:", mean_val)
print("Population Variance:", variance_val)


# ============================================================
# 6. Finding Standard Deviation Using Pure Python
print("\n6. FINDING STANDARD DEVIATION")

data = [2, 4, 4, 4, 5, 5, 7, 9]

mean_val = sum(data) / len(data)

variance_val = sum(
    (x - mean_val) ** 2 for x in data
) / len(data)

std_dev = variance_val ** 0.5

print("Data:", data)
print("Mean:", mean_val)
print("Variance:", variance_val)
print("Standard Deviation:", std_dev)


# ============================================================
# 7. Using Built-in Statistics Module
print("\n7. USING STATISTICS MODULE")

import statistics

data = [12, 15, 15, 18, 22, 25]

print("Data:", data)
print("Mean:", statistics.mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.mode(data))
print("Sample Variance:", statistics.variance(data))
print("Sample Standard Deviation:", statistics.stdev(data))


# ============================================================
# 8. Using NumPy Library
print("\n8. USING NUMPY")

import numpy as np

data = np.array([10, 20, 20, 30, 40])

print("Data:", data)
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Sample Variance:", np.var(data, ddof=1))
print("Sample Standard Deviation:", np.std(data, ddof=1))
print("Range:", np.ptp(data))


# ============================================================
# 9. Using SciPy for Mode
print("\n9. USING SCIPY")

from scipy import stats

data = [5, 2, 5, 3, 5, 4]

mode_result = stats.mode(data, keepdims=True)

print("Data:", data)
print("Mode:", mode_result.mode[0])
print("Count:", mode_result.count[0])


# ============================================================
# 10. Using Pandas for Complete Summary
print("\n10. USING PANDAS")

import pandas as pd

df = pd.DataFrame({
    'values': [10, 20, 30, 40, 50, 50]
})

print("Data:")
print(df)

print("\nMean:", df['values'].mean())
print("Median:", df['values'].median())
print("Mode:", df['values'].mode().tolist())
print("Sample Variance:", df['values'].var())
print("Sample Standard Deviation:", df['values'].std())
print("Range:", df['values'].max() - df['values'].min())
```
