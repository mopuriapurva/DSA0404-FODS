```python
print("=" * 50)
print("1. Mean (Pure Python)")
print("=" * 50)
data = [10, 20, 30, 40, 50]
mean_val = sum(data) / len(data)
print("Data:", data)
print("Mean:", mean_val)
print("\n" + "=" * 50)
print("2. Median (Pure Python)")
print("=" * 50)
data = [40, 10, 30, 50, 20]
data.sort()
n = len(data)
if n % 2 == 0:
    median_val = (data[n//2 - 1] + data[n//2]) / 2
else:
    median_val = data[n//2]

print("Sorted Data:", data)
print("Median:", median_val)


print("\n" + "=" * 50)
print("3. Mode (Pure Python)")
print("=" * 50)

from collections import Counter

data = [1, 2, 2, 3, 4, 2, 5]

count = Counter(data)
mode_val = count.most_common(1)[0][0]

print("Data:", data)
print("Mode:", mode_val)


print("\n" + "=" * 50)
print("4. Range (Pure Python)")
print("=" * 50)

data = [15, 8, 22, 42, 10]

range_val = max(data) - min(data)

print("Data:", data)
print("Range:", range_val)


print("\n" + "=" * 50)
print("5. Variance (Population)")
print("=" * 50)

data = [2, 4, 4, 4, 5, 5, 7, 9]

mean_val = sum(data) / len(data)

variance_val = sum((x - mean_val) ** 2 for x in data) / len(data)

print("Data:", data)
print("Variance:", variance_val)


print("\n" + "=" * 50)
print("6. Standard Deviation (Pure Python)")
print("=" * 50)

std_dev = variance_val ** 0.5

print("Data:", data)
print("Standard Deviation:", std_dev)


print("\n" + "=" * 50)
print("7. Using statistics Module")
print("=" * 50)

import statistics

data = [12, 15, 15, 18, 22, 25]

print("Data:", data)
print("Mean:", statistics.mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.mode(data))
print("Sample Variance:", statistics.variance(data))
print("Sample Standard Deviation:", statistics.stdev(data))


print("\n" + "=" * 50)
print("8. Using NumPy")
print("=" * 50)

import numpy as np

data = np.array([10, 20, 20, 30, 40])

print("Data:", data)
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Sample Variance:", np.var(data, ddof=1))
print("Sample Standard Deviation:", np.std(data, ddof=1))
print("Range:", np.ptp(data))


print("\n" + "=" * 50)
print("9. Using SciPy for Mode")
print("=" * 50)

from scipy import stats

data = [5, 2, 5, 3, 5, 4]

mode_result = stats.mode(data, keepdims=True)

print("Data:", data)
print("Mode:", mode_result.mode[0])
print("Count:", mode_result.count[0])


print("\n" + "=" * 50)
print("10. Using Pandas")
print("=" * 50)

import pandas as pd

df = pd.DataFrame({
    "Values": [10, 20, 30, 40, 50, 50]
})

print(df)

print("Mean:", df["Values"].mean())
print("Median:", df["Values"].median())
print("Mode:", df["Values"].mode().tolist())
print("Sample Variance:", df["Values"].var())
print("Sample Standard Deviation:", df["Values"].std())
print("Range:", df["Values"].max() - df["Values"].min())

print("\nAll programs executed successfully!")
```
